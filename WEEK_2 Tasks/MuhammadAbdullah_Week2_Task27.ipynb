### 1. Conceptual Background
In enterprise and AI applications, standard exceptions are often not expressive enough to indicate domain-specific logic failures. Custom exceptions can be created by defining a new class that inherits directly or indirectly from Python's built-in `Exception` base class.

```
# Defining custom exception hierarchy
class BankAccountError(Exception):
    """Base exception for bank account operations."""
    pass

class InsufficientFundsError(BankAccountError):
    """Raised when withdrawal exceeds available balance."""
    def __init__(self, balance, amount):
        self.balance = balance
        self.amount = amount
        super().__init__(f'Cannot withdraw ${amount:.2f}. Available balance is only ${balance:.2f}.')

class NegativeAmountError(BankAccountError):
    """Raised when a negative transaction amount is submitted."""
    def __init__(self, amount):
        super().__init__(f'Invalid transaction amount: ${amount:.2f}. Amount must be positive.')
```

### 2. Implementing Domain Logic with Custom Exceptions
Demonstrating a BankAccount class raising and catching custom errors.

```
class BankAccount:
    def __init__(self, account_holder, initial_balance=0.0):
        self.account_holder = account_holder
        self.balance = initial_balance

    def withdraw(self, amount):
        if amount <= 0:
            raise NegativeAmountError(amount)
        if amount > self.balance:
            raise InsufficientFundsError(self.balance, amount)
        self.balance -= amount
        return self.balance

account = BankAccount('Muhammad Abdullah', 500.0)

# Scenario 1: Valid withdrawal
print(f'Successful withdrawal: New Balance = ${account.withdraw(150.0):.2f}')

# Scenario 2: Insufficient funds
try:
    account.withdraw(1000.0)
except InsufficientFundsError as e:
    print(f'Caught Custom Exception: {e}')

# Scenario 3: Negative amount
try:
    account.withdraw(-50.0)
except NegativeAmountError as e:
    print(f'Caught Custom Exception: {e}')
```
