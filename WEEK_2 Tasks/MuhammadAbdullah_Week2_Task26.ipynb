### 1. Conceptual Background
Exceptions are runtime errors detected during program execution. Python provides robust exception handling blocks:
- `try`: Code that may raise an error.
- `except`: Code executed if an exception occurs.
- `else`: Code executed if no exceptions occurred.
- `finally`: Cleanup code that **always** executes regardless of exceptions.

```
def safe_division(numerator, denominator):
    print(f'Dividing {numerator} by {denominator}:')
    try:
        result = numerator / denominator
    except ZeroDivisionError as e:
        print(f'  [EXCEPT] Caught ZeroDivisionError: {e}')
    except TypeError as e:
        print(f'  [EXCEPT] Caught TypeError: {e}')
    else:
        print(f'  [ELSE] Division successful! Result = {result:.4f}')
    finally:
        print('  [FINALLY] Operation clean-up finished.\n')

safe_division(50, 5)
safe_division(50, 0)
safe_division(50, 'ten')
```

### 2. Handling File and Input Exceptions
Safely attempting to open non-existent files and parsing numeric strings.

```
try:
    with open('non_existent_data.txt', 'r') as f:
        content = f.read()
except FileNotFoundError as err:
    print(f'Handled Expected Missing File: {err}')

def parse_integer(val):
    try:
        return int(val)
    except ValueError:
        return None

print(f'Parsed valid: {parse_integer("123")}')
print(f'Parsed invalid: {parse_integer("abc12")}')
```
