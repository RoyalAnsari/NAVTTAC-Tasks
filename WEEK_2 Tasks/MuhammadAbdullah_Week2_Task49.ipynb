### 1. Conceptual Background
Measures of central tendency represent the center point or typical value of a dataset:
- **Mean (Arithmetic Average):** $\bar{x} = \frac{1}{n} \sum_{i=1}^{n} x_i$
- **Median:** The middle value when data points are sorted. Robust to outliers.
- **Mode:** The most frequently occurring value in the dataset.

```
import numpy as np
import statistics
from scipy import stats

# Sample dataset: Speed of 15 cars
speed = [99, 86, 87, 88, 111, 86, 103, 87, 94, 78, 77, 85, 86, 85, 86]
print(f'Dataset: {speed}')
print(f'Count of observations: {len(speed)}')

# 1. Mean
mean_val = np.mean(speed)
print(f'Mean Speed: {mean_val:.2f} km/h')

# 2. Median
median_val = np.median(speed)
print(f'Median Speed: {median_val:.2f} km/h')

# 3. Mode
mode_res = stats.mode(speed)
print(f'Mode Speed: {mode_res.mode} (Occurred {mode_res.count} times)')
```

### 2. Sensitivity to Extreme Outliers and Visual Comparison
Demonstrating how an extreme outlier shifts the mean while leaving the median stable, followed by a visualization.

```
import matplotlib.pyplot as plt

salaries = [35000, 38000, 40000, 42000, 45000, 47000, 50000, 52000, 1500000] # outlier CEO salary
print(f'Salaries with Outlier: {salaries}')
print(f'Mean: ${np.mean(salaries):,.2f}')
print(f'Median: ${np.median(salaries):,.2f}')

plt.figure(figsize=(8, 4))
plt.hist(speed, bins=8, color='skyblue', edgecolor='black', alpha=0.7)
plt.axvline(mean_val, color='red', linestyle='--', linewidth=2, label=f'Mean: {mean_val:.1f}')
plt.axvline(median_val, color='green', linestyle='-', linewidth=2, label=f'Median: {median_val:.1f}')
plt.axvline(mode_res.mode, color='purple', linestyle=':', linewidth=2, label=f'Mode: {mode_res.mode}')
plt.title('Vehicle Speed Distribution with Central Tendency Measures')
plt.xlabel('Speed (km/h)')
plt.ylabel('Frequency')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
