### 1. Conceptual Background
A `while` loop repeatedly executes a target block of statements as long as a given boolean condition evaluates to `True`. It is primarily used when the number of iterations is not known beforehand.

```
# Basic While Loop counter
count = 1
print('--- While Loop from 1 to 5 ---')
while count <= 5:
    print(f'Count is: {count}')
    count += 1
```

### 2. Accumulation and Sentinel-Controlled While Loop
Computing the sum of numbers and demonstrating loop convergence.

```
total_sum = 0
num = 1
target = 10

while num <= target:
    total_sum += num
    num += 1

print(f'Sum of integers from 1 to {target} = {total_sum}')
```

### 3. The `while-else` Statement
The `else` suite in a while loop executes when the loop condition becomes `False`.

```
n = 3
while n > 0:
    print(f'Countdown: {n}')
    n -= 1
else:
    print('Countdown complete! Liftoff!')
```
