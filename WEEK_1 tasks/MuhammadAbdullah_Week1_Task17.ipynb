### 1. Conceptual Background
The `break` statement brings control out of the innermost enclosing loop immediately. If an `else` clause is attached to the loop, it is skipped when `break` terminates the loop.

```
# Using break in a for loop
print('--- Break at value 4 ---')
for i in range(1, 10):
    if i == 4:
        print(f'Trigger reached at i = {i}! Breaking out of loop.')
        break
    print(f'Current i: {i}')
print('Loop ended.')
```

### 2. Linear Search Algorithm with Early Termination
Practical demonstration: searching for a target in an unsorted list and breaking immediately upon finding it.

```
records = ['Database', 'Network', 'Security', 'Artificial Intelligence', 'Cloud', 'DevOps']
target = 'Artificial Intelligence'

for index, item in enumerate(records):
    if item == target:
        print(f'Target "{target}" found at index {index}! Halting search.')
        break
else:
    print(f'Target "{target}" was not found in records.')
```

### 3. Breaking out of While Loop
Using break to safely exit an infinite while loop upon meeting a threshold condition.

```
num = 1
while True:
    if num ** 2 > 50:
        print(f'First integer whose square exceeds 50 is: {num} (Square: {num**2})')
        break
    num += 1
```
