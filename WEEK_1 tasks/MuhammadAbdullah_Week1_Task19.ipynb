### 1. Conceptual Background
Python provides elegant, optimized built-in functions for clean and expressive iteration: `enumerate()`, `zip()`, `reversed()`, `sorted()`, and dictionary `.items()`.

```
# 1. enumerate() - Tracking element indices cleanly
names = ['Python', 'PyTorch', 'TensorFlow', 'Scikit-Learn']
print('--- enumerate() ---')
for idx, framework in enumerate(names, start=1):
    print(f'{idx}. Framework: {framework}')

# 2. zip() - Pairing multiple sequences simultaneously
students = ['Abdullah', 'Zain', 'Hamza']
scores = [94, 88, 76]
cities = ['Lahore', 'Karachi', 'Islamabad']
print('\n--- zip() ---')
for student, score, city in zip(students, scores, cities):
    print(f'{student:10s} scored {score:2d} from {city}')
```

### 2. `reversed()`, `sorted()`, and List Comprehensions
Demonstrating iteration modifiers and compact list comprehensions.

```
numbers = [14, 3, 27, 8, 42, 19]

# sorted() looping
print('--- sorted() iteration ---')
for n in sorted(numbers):
    print(n, end=' ')
print()

# reversed() looping
print('--- reversed() iteration ---')
for n in reversed(numbers):
    print(n, end=' ')
print()

# List comprehension with condition
squares_of_evens = [x**2 for x in numbers if x % 2 == 0]
print(f'Squares of even numbers: {squares_of_evens}')
```
