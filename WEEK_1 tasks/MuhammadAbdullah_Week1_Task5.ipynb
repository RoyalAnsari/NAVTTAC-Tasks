### 1. Conceptual Background
In mathematics, the square root of $x$ is equal to $x^{0.5}$ or $x^{1/2}$. Python provides the exponentiation operator `**` and the standard `math.sqrt()` function.

```
# Python Program to calculate the square root
# Note: change this value for a different result
num = 8

# Calculate square root using exponent operator
num_sqrt = num ** 0.5

print('The square root of %0.3f is %0.3f' % (num, num_sqrt))
```

### 2. Comparison with `math.sqrt()` and Multiple Test Cases
Testing multiple numbers and verifying output against the Python standard library.

```
import math

test_numbers = [4, 9, 16, 25, 50, 81, 144]
print('Number | Exponent ** 0.5 | math.sqrt() | Match?')
print('-' * 45)
for n in test_numbers:
    r1 = n ** 0.5
    r2 = math.sqrt(n)
    print(f'{n:6d} | {r1:15.4f} | {r2:11.4f} | {math.isclose(r1, r2)}')
```
