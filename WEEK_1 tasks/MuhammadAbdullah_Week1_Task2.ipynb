### 2. Comprehensive Linux Commands Reference Table

| Command | Category | Description & Usage Example |
| :--- | :--- | :--- |
| `pwd` | Navigation | **Print Working Directory**: outputs current absolute path (`pwd`) |
| `cd` | Navigation | **Change Directory**: navigate file hierarchy (`cd /var/log`, `cd ..`, `cd ~`) |
| `ls` | Inspection | **List Directory Contents**: display directory files (`ls -la`, `ls -lh`) |
| `mkdir` | File Ops | **Make Directory**: create new folders (`mkdir -p models/checkpoints`) |
| `touch` | File Ops | **Create File / Update Timestamp**: (`touch train.py`, `touch dataset.csv`) |
| `cp` | File Ops | **Copy**: copy files or directories recursively (`cp -r src/ dest/`) |
| `mv` | File Ops | **Move / Rename**: relocate or rename files (`mv old.txt new.txt`) |
| `rm` | File Ops | **Remove**: delete files or directories (`rm -rf temp_dir/`) |
| `rmdir` | File Ops | **Remove Empty Directory**: deletes empty folder (`rmdir empty_dir`) |
| `cat` | Viewing | **Concatenate & Display**: view full file content (`cat config.yaml`) |
| `more` | Viewing | **Pager**: view file contents screen-by-screen (`more large_log.txt`) |
| `less` | Viewing | **Advanced Pager**: scroll backward and forward through large files (`less data.log`) |
| `head` | Viewing | **View Top Lines**: output first $n$ lines of a file (`head -n 10 file.csv`) |
| `file` | Inspection | **Determine File Type**: inspect MIME/binary type (`file weights.pt`) |
| `redirection` | I/O Ops | **I/O Redirection**: `>` overwrite, `>>` append, `<` input, `|` pipe (`ls | grep py`) |
| `nano` | Editors | **Terminal Text Editor**: easy in-terminal editing (`nano script.sh`) |
| `bash` / `sh` | Shell | **Bourne-Again Shell / Shell**: execute scripts (`bash run.sh`, `sh build.sh`) |
| `chmod` | Security | **Change File Permissions**: (`chmod +x train.sh`, `chmod 755 app.py`) |
| `sudo` | Admin | **Superuser Do**: execute command with administrative root privileges (`sudo apt update`) |
| `man` | Help | **Manual Pages**: display command documentation (`man ls`, `man grep`) |
| `alias` | Shell Ops | **Command Shortcut**: define custom shortcuts (`alias py=python3`) |
| `ps` | Process | **Process Status**: snapshot of currently running processes (`ps aux | grep python`) |
| `kill` | Process | **Terminate Process**: send signals to PID (`kill -9 1234`) |
| `dpkg` | Packages | **Debian Package Manager**: inspect and install `.deb` packages (`dpkg -l`) |
| `shutdown` | System | **Halt System**: safely shut down machine (`sudo shutdown -h now`) |
| `restart` | System | **Reboot System**: restart operating system (`sudo reboot` / `sudo shutdown -r now`) |

### 3. Practical Python Simulation of Core Linux Operations
Below we demonstrate filesystem operations corresponding directly to `mkdir`, `touch`, `cat`, `head`, `cp`, `mv`, `ls`, and `rm` within a sandbox directory.

```
import os
import shutil
import glob

sandbox = 'linux_sandbox'
os.makedirs(sandbox, exist_ok=True)
print(f'[mkdir] Created sandbox directory: {sandbox}')

# [touch] & writing content
sample_file = os.path.join(sandbox, 'dataset_info.txt')
with open(sample_file, 'w', encoding='utf-8') as f:
    f.write('Dataset: MNIST Handwritten Digits\n')
    f.write('Samples: 70,000 images\n')
    f.write('Image Size: 28x28 grayscale\n')
    f.write('Classes: 10 (digits 0 to 9)\n')
    f.write('Format: NumPy arrays\n')
print(f'[touch] Created and populated: {sample_file}')

# [cat] Reading entire file
print('\n--- [cat] Full File Contents ---')
with open(sample_file, 'r') as f:
    print(f.read().strip())

# [head] Reading first 2 lines
print('\n--- [head -n 2] First 2 Lines ---')
with open(sample_file, 'r') as f:
    for _ in range(2):
        print(f.readline().strip())

# [cp] Copying file
copied_file = os.path.join(sandbox, 'dataset_backup.txt')
shutil.copy(sample_file, copied_file)
print(f'\n[cp] Copied file to: {copied_file}')

# [ls] Listing directory contents
print(f'\n[ls] Files in {sandbox}: {os.listdir(sandbox)}')

# Clean up sandbox directory (rm -rf)
shutil.rmtree(sandbox)
print(f'[rm -rf] Successfully cleaned up {sandbox}')
```

### 4. Process & Permissions Simulation (`ps`, `chmod`)
Demonstrating process identification and permission mask interpretation.

```
import os
import stat

# [ps] Current process ID
print(f'[ps] Current Process ID (PID): {os.getpid()}')
if hasattr(os, 'getppid'):
    print(f'[ps] Parent Process ID (PPID):  {os.getppid()}')

# [chmod] Demonstrating symbolic & octal permission bits
permissions_map = {
    '755': 'rwxr-xr-x (Owner: Read/Write/Exec | Group: Read/Exec | Others: Read/Exec)',
    '644': 'rw-r--r-- (Owner: Read/Write | Group: Read | Others: Read)',
    '700': 'rwx------ (Owner: Read/Write/Exec | Private to Owner)',
    '777': 'rwxrwxrwx (Full permissions for everyone - avoid in production)'
}

print('--- [chmod] Octal Permission Cheat Sheet ---')
for octal, desc in permissions_map.items():
    print(f'chmod {octal} -> {desc}')
```
