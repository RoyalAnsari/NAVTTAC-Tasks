### 1. Conceptual Background
A list in Python is an ordered, mutable sequence of items. Lists can store heterogeneous data types (integers, strings, booleans, sub-lists) and support dynamic resizing.

```
# Creating lists
numbers = [10, 20, 30, 40, 50]
mixed = [1, 'NAVTTC', 3.14, True]
print(f'Numbers List: {numbers}')
print(f'Mixed Types List: {mixed}')

# Indexing and Slicing
print(f'First Element (index 0): {numbers[0]}')
print(f'Last Element (index -1): {numbers[-1]}')
print(f'Slice [1:4]: {numbers[1:4]}')
print(f'Reverse Slice [::-1]: {numbers[::-1]}')
```

### 2. Modifying Lists: Adding, Inserting, and Removing Elements
Demonstrating methods for mutating lists in place.

```
courses = ['Python', 'Machine Learning']
print(f'Initial List: {courses}')

# append, extend, insert
courses.append('Deep Learning')
courses.extend(['Computer Vision', 'NLP'])
courses.insert(1, 'Data Science')
print(f'After Additions: {courses}')

# remove, pop
removed_item = courses.pop(2)  # removes item at index 2
print(f'Popped item: {removed_item}')
courses.remove('NLP')
print(f'After Removals: {courses}')
```

### 3. List Sorting, Aggregation, and Comprehensions
Using built-in methods and list comprehensions.

```
values = [23, 7, 89, 45, 12, 56]
print(f'Length: {len(values)}, Min: {min(values)}, Max: {max(values)}, Sum: {sum(values)}')

values.sort()
print(f'Sorted In-Place: {values}')

# List comprehension: square of each number greater than 20
squared_filtered = [v**2 for v in values if v > 20]
print(f'Filtered Squares: {squared_filtered}')
```
