### 1. Conceptual Background
When computing square roots of negative numbers or complex numbers ($a + bi$), the standard `math` module raises a `ValueError`. Python's `cmath` (complex math) module supports complex arithmetic seamlessly.

```
# Find square root of real or complex numbers
# Importing the complex math module
import cmath

num = 1 + 2j

# Calculate square root
num_sqrt = cmath.sqrt(num)

print('The square root of {0} is {1:0.3f}+{2:0.3f}j'.format(num, num_sqrt.real, num_sqrt.imag))
```

### 2. Square Root of Negative Real Numbers
Demonstrating square roots of negative numbers like $-16$ and $-49$.

```
negative_nums = [-4, -9, -16, -25]
for n in negative_nums:
    sqrt_res = cmath.sqrt(n)
    print(f'sqrt({n}) = {sqrt_res.real:.1f} + {sqrt_res.imag:.1f}j')
```
