### 1. Conceptual Background
A `for` loop in Python iterates over the items of any sequence (such as a list, tuple, string, or range) in the order that they appear. Unlike index-based loops in C/C++, Python's `for` loop is a collection iterator.

```
# Basic range iteration
print('--- Iterating with range(5) ---')
for i in range(5):
    print(f'Iteration index: {i}')

# Range with start, stop, step
print('\n--- Iterating with range(10, 30, 5) ---')
for val in range(10, 30, 5):
    print(f'Value: {val}')
```

### 2. Iterating Over Strings, Lists, and Dictionaries
Demonstrating iteration over diverse standard Python collections.

```
fruits = ['Apple', 'Banana', 'Cherry', 'Mango']
print('--- Iterating over List ---')
for fruit in fruits:
    print(f'Fruit: {fruit}')

student_scores = {'Ali': 85, 'Sara': 92, 'Usman': 78}
print('\n--- Iterating over Dictionary Items ---')
for student, score in student_scores.items():
    print(f'Student: {student:6s} | Score: {score}')
```

### 3. Nested Loops and the `for-else` Construct
In Python, an optional `else` block executes after the loop completes normally without encountering a `break`.

```
# Nested for loop (Multiplication Table Grid)
print('--- 3x3 Multiplication Grid ---')
for row in range(1, 4):
    for col in range(1, 4):
        print(f'{row * col:3d}', end=' ')
    print()

# For-Else demonstrating complete iteration
print('\n--- For-Else Demonstration ---')
for n in [2, 4, 6]:
    print(f'Processing {n}')
else:
    print('Loop completed all iterations successfully with no break.')
```
