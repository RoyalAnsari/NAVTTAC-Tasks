### 1. Conceptual Background
Nested `if` statements occur when an `if` or `else` block contains one or more additional `if` statements. This allows modeling multi-step dependent decision logic.

```
i = 10

if (i == 10):
    # First if statement
    if (i < 15):
        print('i is smaller than 15')
    # Nested if statement
    # Will only be executed if statement above it is true
    if (i < 12):
        print('i is smaller than 12 too')
    else:
        print('i is greater than 15')
```

### 2. Practical Use Case: Student Admission Eligibility
Evaluating multiple dependent conditions (marks and interview score) with nested conditionals.

```
def check_eligibility(marks, interview_passed):
    print(f'Candidate -> Marks: {marks}, Interview: {interview_passed}')
    if marks >= 60:
        if interview_passed:
            print('  -> Result: Fully Admitted with Scholarship.')
        else:
            print('  -> Result: Provisionally Admitted (Interview Retake required).')
    else:
        print('  -> Result: Not Eligible based on marks criteria.')

check_eligibility(85, True)
check_eligibility(72, False)
check_eligibility(45, True)
```
