### 1. Conceptual Background
Heron's formula allows finding the area of any triangle when the lengths of all three sides ($a, b, c$) are known:
1. Semi-perimeter: $s = \frac{a + b + c}{2}$
2. Area: $A = \sqrt{s(s - a)(s - b)(s - c)}$

```
# Python Program to find the area of triangle
a = 5
b = 6
c = 7

# calculate the semi-perimeter
s = (a + b + c) / 2

# calculate the area
area = (s * (s - a) * (s - b) * (s - c)) ** 0.5

print('The area of the triangle is %0.2f' % area)
```

### 2. Validation with Triangle Inequality Theorem
A triangle is valid if and only if the sum of any two sides is strictly greater than the third side: $a+b > c$, $a+c > b$, $b+c > a$.

```
def compute_triangle_area(a, b, c):
    if (a + b > c) and (a + c > b) and (b + c > a):
        s = (a + b + c) / 2
        area = (s * (s - a) * (s - b) * (s - c)) ** 0.5
        return f'Sides ({a}, {b}, {c}) -> Valid Triangle -> Area = {area:.2f}'
    else:
        return f'Sides ({a}, {b}, {c}) -> Invalid Triangle (violates triangle inequality)'

print(compute_triangle_area(3, 4, 5))    # Right angled triangle
print(compute_triangle_area(5, 6, 7))    # Scalene triangle
print(compute_triangle_area(1, 2, 10))   # Invalid triangle
```
