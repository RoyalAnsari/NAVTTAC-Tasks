### 1. Conceptual Background
Strings in Python are immutable sequences of Unicode characters. They support negative indexing, slicing, concatenation, and an extensive collection of built-in transformation methods.

```
text = 'NAVTTC AI Specialization 2026'
print(f'Text: {text}')
print(f'Length: {len(text)}')
print(f'Substring (0:6): "{text[0:6]}"')
print(f'Reverse string: "{text[::-1]}"')
```

### 2. Common Built-in String Methods
Case transformations, trimming whitespace, searching, and replacing substrings.

```
raw_str = '   Machine Learning and Deep Learning with Python   '
clean_str = raw_str.strip()
print(f'Stripped: "{clean_str}"')
print(f'Upper Case: {clean_str.upper()}')
print(f'Lower Case: {clean_str.lower()}')
print(f'Title Case: {clean_str.title()}')
print(f'Replaced "Python" with "PyTorch": {clean_str.replace("Python", "PyTorch")}')
print(f'Find position of "Deep": {clean_str.find("Deep")}')
```

### 3. Splitting, Joining, and Formatted Strings
Tokenization into lists using `.split()` and reassembly using `.join()`.

```
sentence = 'Data Science, Machine Learning, Deep Learning, Generative AI'
domains = sentence.split(', ')
print(f'Split into list: {domains}')

rejoined = ' -> '.join(domains)
print(f'Rejoined with delimiter: {rejoined}')

# Advanced f-string formatting
accuracy = 0.98457
epoch = 25
print(f'Epoch: {epoch:03d} | Model Validation Accuracy: {accuracy * 100:.2f}%')
```
