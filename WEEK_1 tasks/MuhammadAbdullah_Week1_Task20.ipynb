### 1. Conceptual Background
A function is a block of organized, reusable code used to perform a single, related action. Functions provide better modularity for applications and a high degree of code reusability. They are defined using the `def` keyword followed by the function name and parentheses `()`.

```python
def greet_student(name, course='Artificial Intelligence'):
    """Docstring: Greets a student and displays their enrolled course."""
    return f'Welcome, {name}! Enrolled in: {course}'

# Calling function with positional and default arguments
print(greet_student('Muhammad Abdullah'))
print(greet_student('Sara Khan', 'Data Science & ML'))
```

### 2. Variable-Length Arguments: `*args` and `**kwargs`
`*args` accepts an arbitrary number of positional arguments as a tuple, while `**kwargs` accepts arbitrary keyword arguments as a dictionary.

```python
def compute_stats(*scores, **metadata):
    avg_score = sum(scores) / len(scores) if scores else 0
    print('--- Student Report ---')
    for key, value in metadata.items():
        print(f'{key.capitalize()}: {value}')
    print(f'Total Subjects: {len(scores)}')
    print(f'Average Score: {avg_score:.2f}')
    return avg_score

compute_stats(85, 90, 78, 92, student='Abdullah', program='NAVTTC AI', semester='Fall 2026')
```

### 3. Recursive Functions and Lambda Expressions
Demonstrating recursion (factorial) and anonymous lambda functions with `map` and `filter`.

```python
# Recursive factorial
def factorial(n):
    if n <= 1:
        return 1
    return n * factorial(n - 1)

print(f'Factorial of 5: {factorial(5)}')

# Lambda function
square = lambda x: x ** 2
nums = [1, 2, 3, 4, 5, 6]
evens = list(filter(lambda x: x % 2 == 0, nums))
squared_evens = list(map(square, evens))
print(f'Original: {nums}')
print(f'Filtered Evens: {evens}')
print(f'Squared Evens: {squared_evens}')
```
