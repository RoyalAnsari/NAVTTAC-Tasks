### 1. Conceptual Background
A quadratic equation has the general form $ax^2 + bx + c = 0$ (where $a \neq 0$).
The roots are determined by the quadratic formula:
$$x = \frac{-b \pm \sqrt{b^2 - 4ac}}{2a}$$
The discriminant $\Delta = b^2 - 4ac$ dictates the nature of the roots:
- $\Delta > 0$: Two distinct real roots
- $\Delta = 0$: One repeated real root
- $\Delta < 0$: Two complex conjugate roots

```
# Solve the quadratic equation ax**2 + bx + c = 0
# import complex math module
import cmath

a = 1
b = 5
c = 6

# calculate the discriminant
d = (b ** 2) - (4 * a * c)

# find two solutions
sol1 = (-b - cmath.sqrt(d)) / (2 * a)
sol2 = (-b + cmath.sqrt(d)) / (2 * a)

print('The solutions are {0} and {1}'.format(sol1, sol2))
```

### 2. General Quadratic Equation Solver
Demonstrating all three discriminant conditions: positive, zero, and negative.

```
def solve_quadratic(a, b, c):
    d = (b ** 2) - (4 * a * c)
    sol1 = (-b - cmath.sqrt(d)) / (2 * a)
    sol2 = (-b + cmath.sqrt(d)) / (2 * a)
    return sol1, sol2, d

cases = [
    (1, 5, 6),    # Real distinct roots
    (1, -4, 4),   # Repeated real roots
    (1, 2, 5)     # Complex conjugate roots
]

for a, b, c in cases:
    r1, r2, d = solve_quadratic(a, b, c)
    print(f'Equation: {a}x^2 + {b}x + {c} = 0 | Discriminant: {d}')
    print(f'  Root 1: {r1}')
    print(f'  Root 2: {r2}\n')
```
