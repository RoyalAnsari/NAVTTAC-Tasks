### 1. Conceptual Background
The `if` statement is the most fundamental decision-making statement in Python. It tests a boolean expression: if the expression evaluates to `True`, the indented block of code executes; otherwise, the block is bypassed.

```
# Initial value of i
i = 10

if (i > 15):
    print('10 is less than 15')

print('I am Not in if')
```

### 2. Contrasting True vs False Branch Execution
Demonstrating what happens when the condition evaluates to True versus False.

```
def test_condition(val, threshold=15):
    print(f'Testing value val = {val} against threshold {threshold}:')
    if val < threshold:
        print(f'  [INSIDE IF]: {val} is strictly less than {threshold}')
    print('  [OUTSIDE IF]: Execution flow resumes sequentially.\n')

test_condition(10)
test_condition(20)
```
