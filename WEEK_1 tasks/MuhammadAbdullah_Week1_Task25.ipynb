### 1. Conceptual Background
A dictionary in Python is an associative data structure storing data as key-value pairs (`{key: value}`). Keys must be immutable and unique, while values can be of any data type. Dictionaries provide $O(1)$ average time complexity for lookups.

```
# Creating dictionaries
student = {
    'name': 'Muhammad Abdullah',
    'roll_no': 'NIAI_00004',
    'course': 'Artificial Intelligence',
    'marks': [88, 92, 95]
}
print('Student Dictionary:')
for k, v in student.items():
    print(f'  {k:10s} : {v}')

# Accessing values safely with get()
print(f'Name: {student["name"]}')
print(f'GPA (default 4.0 if not present): {student.get("gpa", 4.0)}')
```

### 2. Modifying, Updating, and Removing Dictionary Items
Demonstrating in-place updates, adding new keys, and deletion.

```
# Adding and modifying
student['gpa'] = 3.95
student['course'] = 'AI & Deep Learning Specialization'
print(f'Updated Student: {student}')

# Removing entries with pop and del
removed_marks = student.pop('marks')
print(f'Popped marks: {removed_marks}')
print(f'Remaining keys: {list(student.keys())}')
```

### 3. Dictionary Comprehension and Nested Dictionaries
Creating structured dictionaries concisely.

```
# Dictionary comprehension
numbers = [1, 2, 3, 4, 5]
cubes_dict = {x: x**3 for x in numbers}
print(f'Numbers and Cubes: {cubes_dict}')

# Nested dictionary
classroom = {
    'student_1': {'name': 'Abdullah', 'grade': 'A+'},
    'student_2': {'name': 'Ali', 'grade': 'A'}
}
print(f'Classroom: {classroom}')
print(f'Student 1 Name: {classroom["student_1"]["name"]}')
```
