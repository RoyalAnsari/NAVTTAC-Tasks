### 1. Conceptual Background & Toolchain Overview
Setting up a robust machine learning and deep learning development environment requires several core tools:

1. **Anaconda3:** The leading open-source distribution for scientific Python. It bundles Python with `conda`, a package and environment management system that prevents library conflicts.
2. **PyTorch:** Developed by Meta AI, PyTorch provides dynamic computation graphs (`Autograd`) and GPU-accelerated tensor computing.
3. **TensorFlow 2.0:** Developed by Google Brain, TensorFlow offers a comprehensive ecosystem with high-level Keras APIs for production model deployment.
4. **Visual Studio Code (VS Code):** A lightweight, extensible code editor with rich Jupyter notebook and Python extensions.
5. **PyCharm:** A full-featured Python IDE developed by JetBrains, offering deep code inspection, refactoring, and debugging.

### 2. Comprehensive Installation Guide & CLI Commands

#### A. Anaconda3 Installation
- Download installer from: `https://www.anaconda.com/download`
- Create a dedicated virtual environment:
```bash
conda create -n navttc_ai python=3.10 -y
conda activate navttc_ai
```

#### B. PyTorch Installation


```bash
pip install torch torchvision torchaudio
```


### 3. Environment & Package Verification Script
The following Python code verifies the active runtime, operating system, and installed AI libraries.

````python
import sys
import os
import platform

print('=== SYSTEM & RUNTIME VERIFICATION ===')
print(f'Python Version:       {platform.python_version()}')
print(f'Python Executable:    {sys.executable}')
print(f'Operating System:     {platform.system()} {platform.release()} ({platform.architecture()[0]})')
print(f'Platform Machine:     {platform.machine()}')
print(f'Current Directory:    {os.getcwd()}')

# Verifying core scientific AI stack
packages = [
    'numpy',
    'pandas',
    'scipy',
    'sklearn',
    'matplotlib'
]

print('\n=== INSTALLED CORE AI PACKAGES ===')
print(f'{"Package Name":<15} | {"Status":<12} | {"Version"}')
print('-' * 45)
for pkg in packages:
    try:
        mod = __import__(pkg)
        version = getattr(mod, '__version__', 'Installed')
        print(f'{pkg:<15} | {"Available":<12} | {version}')
    except ImportError:
        print(f'{pkg:<15} | {"Missing":<12} | N/A')
````

````python
print('=== DEEP LEARNING FRAMEWORK STATUS ===')

# PyTorch check
try:
    import torch
    cuda_avail = torch.cuda.is_available()
    print(f'PyTorch Version:     {torch.__version__}')
    print(f'CUDA (GPU) Support:  {cuda_avail}')
    if cuda_avail:
        print(f'GPU Device Name:     {torch.cuda.get_device_name(0)}')
except ImportError:
    print('PyTorch:              Not installed in this environment (Ready for: pip install torch)')

# TensorFlow check
try:
    import tensorflow as tf
    print(f'TensorFlow Version:  {tf.__version__}')
    gpus = tf.config.list_physical_devices('GPU')
    print(f'TensorFlow GPUs:     {len(gpus)}')
except ImportError:
    print('TensorFlow:           Not installed in this environment (Ready for: pip install tensorflow)')

print('\nAll installation requirements for Task 1 have been documented and verified.')
````
