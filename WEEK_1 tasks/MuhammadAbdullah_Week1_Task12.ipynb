### 1. Conceptual Background
The `if-else` construct provides binary branching: if the specified condition is True, the `if` block executes; otherwise, control jumps directly to the `else` block.

```
i = 20

if (i < 15):
    print('i is smaller than 15')
    print("i'm in if Block")
else:
    print('i is greater than 15')
    print("i'm in else Block")

print("i'm not in if and not in else Block")
```

### 2. Practical Application: Even vs Odd and Pass vs Fail
Applying if-else logic in real-world programming contexts.

```
numbers = [12, 17, 24, 35, 40]
for num in numbers:
    if num % 2 == 0:
        status = 'Even'
    else:
        status = 'Odd'
    print(f'Number {num:2d} is {status}')
```
