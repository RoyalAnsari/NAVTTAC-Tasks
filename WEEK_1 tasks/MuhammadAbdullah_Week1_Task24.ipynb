### 1. Conceptual Background
A set in Python is an unordered collection of unique, hashable items. Sets automatically eliminate duplicate elements and support mathematical set operations like union, intersection, and difference.

```
# Creating sets
my_set = {10, 20, 30, 20, 10, 40, 50}
print(f'Set with automatic deduplication: {my_set}')

# Removing duplicates from a list using set
raw_numbers = [1, 2, 2, 3, 4, 4, 4, 5, 5]
unique_numbers = list(set(raw_numbers))
print(f'Original List: {raw_numbers}')
print(f'Unique Elements: {unique_numbers}')
```

### 2. Modifying Sets: `add`, `update`, `remove`, `discard`
Contrasting `remove()` (raises KeyError if absent) with `discard()` (silent if absent).

```
languages = {'Python', 'Java', 'C++'}
print(f'Initial Set: {languages}')

languages.add('Julia')
languages.update(['Go', 'Rust'])
print(f'After Add & Update: {languages}')

languages.discard('Java')  # Safe removal
languages.discard('Swift') # No error even if absent
print(f'After Discard: {languages}')
```

### 3. Mathematical Set Operations
Union, Intersection, Difference, and Symmetric Difference.

```
set_A = {1, 2, 3, 4, 5}
set_B = {4, 5, 6, 7, 8}

print(f'Set A: {set_A}')
print(f'Set B: {set_B}')
print(f'Union (A | B): {set_A | set_B}')
print(f'Intersection (A & B): {set_A & set_B}')
print(f'Difference (A - B): {set_A - set_B}')
print(f'Difference (B - A): {set_B - set_A}')
print(f'Symmetric Difference (A ^ B): {set_A ^ set_B}')
```
