### 1. Conceptual Background
Real-world data is frequently messy, containing missing (`NaN`), invalid, or duplicated records. Pandas provides specialized data cleaning functions:
- `dropna()`: Remove rows/columns with null values.
- `fillna(value)`: Impute missing values with constants or statistical measures (mean, median, mode).
- `drop_duplicates()`: Detect and remove redundant duplicate rows.

```
import pandas as pd
import numpy as np

# Constructing a dirty dataset
dirty_data = {
    'ID': [1, 2, 3, 4, 5, 5, 6],
    'Age': [25, np.nan, 30, 22, 28, 28, 35],
    'Salary': [50000, 60000, np.nan, 45000, 52000, 52000, 70000],
    'City': ['Karachi', 'Lahore', 'Islamabad', 'Karachi', 'Lahore', 'Lahore', 'Karachi']
}

df = pd.DataFrame(dirty_data)
print('Raw Dirty Dataset:')
print(df)
```

### 2. Imputing Missing Values and Removing Duplicates
Applying standard cleaning workflows step-by-step.

```
# 1. Check null counts
print('Missing Values Count per Column:')
print(df.isnull().sum())

# 2. Impute missing Age with median age
median_age = df['Age'].median()
df['Age'].fillna(median_age, inplace=True)
print(f'\nImputed missing Age with median ({median_age})')

# 3. Impute missing Salary with mean salary
mean_salary = df['Salary'].mean()
df['Salary'].fillna(mean_salary, inplace=True)
print(f'Imputed missing Salary with mean ({mean_salary:.2f})')

# 4. Detect and drop duplicate records
print(f'Duplicate rows detected: {df.duplicated().sum()}')
df.drop_duplicates(inplace=True)

print('\nCleaned Dataset:')
print(df.reset_index(drop=True))
```
