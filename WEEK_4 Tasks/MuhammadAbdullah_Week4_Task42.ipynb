### 1. Conceptual Background
To use Pandas, it is imported using the industry-standard alias `import pandas as pd`. A DataFrame can be initialized from dictionaries, lists of dictionaries, 2D NumPy arrays, or external files.

```
import pandas as pd

# Creating DataFrame from a dictionary of lists
data = {
    'Student_Name': ['Muhammad Abdullah', 'Ayesha', 'Bilal', 'Zainab'],
    'Score': [95, 89, 78, 92],
    'Grade': ['A+', 'A', 'B', 'A+']
}

df_students = pd.DataFrame(data)
print('--- Students DataFrame ---')
print(df_students)
```

### 2. Inspecting Structural Attributes
Checking dimensions, column names, index range, and memory footprint.

```
print(f'Number of rows: {len(df_students)}')
print(f'Index Range: {df_students.index}')
print(f'Shape tuple: {df_students.shape}')
print(f'Number of elements (size): {df_students.size}')
print(f'Memory Usage:\n{df_students.memory_usage()}')
```
