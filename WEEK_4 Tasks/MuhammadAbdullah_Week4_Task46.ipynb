### 1. Conceptual Background
Before modeling or cleaning, data scientists inspect the dataset using standard diagnostic methods:
- `.head(n)`: Returns the first $n$ rows (default 5).
- `.tail(n)`: Returns the last $n$ rows (default 5).
- `.info()`: Prints concise summary of DataFrame (dtypes, non-null counts, memory).
- `.describe()`: Calculates statistical summaries for numerical columns.

```
import pandas as pd
import numpy as np

# Generating a realistic synthetic customer dataset
np.random.seed(42)
n = 20
df = pd.DataFrame({
    'Customer_ID': range(1001, 1001 + n),
    'Age': np.random.randint(20, 65, size=n),
    'Annual_Income_k': np.random.normal(60, 15, size=n).round(1),
    'Spending_Score': np.random.randint(1, 100, size=n),
    'Loyalty_Member': np.random.choice(['Yes', 'No'], size=n, p=[0.4, 0.6])
})

print('--- First 5 Rows (.head()) ---')
print(df.head())

print('\n--- Last 3 Rows (.tail(3)) ---')
print(df.tail(3))
```

### 2. Concise Dataset Information and Summary Statistics
Using `.info()` and `.describe()`.

```
print('--- Dataset Info (.info()) ---')
df.info()

print('\n--- Descriptive Summary (.describe()) ---')
print(df.describe().round(2))
```
