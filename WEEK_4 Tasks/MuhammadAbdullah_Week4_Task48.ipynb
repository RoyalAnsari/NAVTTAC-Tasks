### 1. Conceptual Background
Correlation measures the linear relationship between two numerical variables. The Pearson correlation coefficient ($r$) ranges between $-1$ and $+1$:
- $r = 1$: Perfect positive linear correlation.
- $r = 0$: No linear relationship.
- $r = -1$: Perfect negative linear correlation.
The `.corr()` method computes pairwise correlations for all numerical columns in a DataFrame.

```
import pandas as pd
import numpy as np

# Sample fitness training dataset
data = {
    'Duration': [60, 60, 60, 45, 45, 60, 60, 45, 30, 60],
    'Pulse': [110, 117, 103, 109, 117, 102, 110, 104, 109, 115],
    'Maxpulse': [130, 145, 135, 175, 148, 127, 136, 134, 133, 137],
    'Calories': [409.1, 479.0, 340.0, 282.4, 406.0, 300.0, 374.0, 253.3, 195.1, 375.0]
}

df = pd.DataFrame(data)
print('Dataset Sample:')
print(df.head())

# Computing correlation matrix
corr_matrix = df.corr()
print('\nPairwise Pearson Correlation Matrix:')
print(corr_matrix.round(4))
```

### 2. Visualizing Correlation Matrix with Heatmap
Rendering a visual matrix using matplotlib.

```
import matplotlib.pyplot as plt

plt.figure(figsize=(7, 5))
plt.imshow(corr_matrix, cmap='coolwarm', vmin=-1, vmax=1)
plt.colorbar(label='Pearson Correlation Coefficient')

cols = corr_matrix.columns
plt.xticks(range(len(cols)), cols, rotation=45)
plt.yticks(range(len(cols)), cols)

# Annotating cell values
for i in range(len(cols)):
    for j in range(len(cols)):
        plt.text(j, i, f'{corr_matrix.iloc[i, j]:.2f}',
                 ha='center', va='center',
                 color='white' if abs(corr_matrix.iloc[i, j]) > 0.5 else 'black')

plt.title('Pairwise Correlation Heatmap')
plt.tight_layout()
plt.show()
```
