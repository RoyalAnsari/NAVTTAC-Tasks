### 1. Conceptual Background
A **DataFrame** is a 2-dimensional labeled data structure with columns of potentially different types. Rows can be accessed by:
- `.loc[]`: Label-based indexing.
- `.iloc[]`: Integer-position based indexing.

```
import pandas as pd

data = {
    'Calories': [420, 380, 390, 450],
    'Duration_Min': [50, 40, 45, 60]
}

df = pd.DataFrame(data, index=['Session 1', 'Session 2', 'Session 3', 'Session 4'])
print('Workout DataFrame with Named Index:')
print(df)
```

### 2. Row and Column Operations: `.loc`, `.iloc`, Adding Columns
Selecting subsets of rows/columns and creating calculated columns.

```
# Using .loc (label)
print(f'Session 1 data using .loc:\n{df.loc["Session 1"]}\n')
# Using .iloc (integer index)
print(f'First two rows using .iloc:\n{df.iloc[:2]}\n')

# Adding calculated column: Calorie burn rate (calories / min)
df['Cal_Per_Min'] = (df['Calories'] / df['Duration_Min']).round(2)
print('DataFrame after adding Cal_Per_Min column:')
print(df)

# Conditional filtering: sessions where calories > 400
high_burn = df[df['Calories'] > 400]
print(f'\nHigh burn sessions (> 400 cal):\n{high_burn}')
```
