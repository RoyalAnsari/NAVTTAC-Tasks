### 1. Conceptual Background
A Pandas **Series** is like a column in a table. It is a one-dimensional array holding data of any type. By default, rows are labeled with integers starting from 0, but custom index labels can be explicitly assigned.

```
import pandas as pd

# Series with default integer index
cal_data = [420, 380, 390]
s1 = pd.Series(cal_data)
print('Series with default integer index:')
print(s1)

# Series with custom index labels
s2 = pd.Series(cal_data, index=['Day 1', 'Day 2', 'Day 3'])
print('\nSeries with custom day labels:')
print(s2)
```

### 2. Creating Series from Dictionaries and Element Lookup
When creating a Series from a dictionary, the keys automatically become the index labels.

```
# Series from dictionary
calories_dict = {'Monday': 2100, 'Tuesday': 2250, 'Wednesday': 1950, 'Thursday': 2300}
cal_series = pd.Series(calories_dict)
print('Series from Dictionary:')
print(cal_series)

# Label-based lookup
print(f'Calories on Wednesday: {cal_series["Wednesday"]}')
# Positional lookup
print(f'First day calories (iloc 0): {cal_series.iloc[0]}')
# Vectorized condition
print(f'Days exceeding 2000 calories:\n{cal_series[cal_series > 2000]}')
```
