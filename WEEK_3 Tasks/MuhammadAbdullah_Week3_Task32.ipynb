### 1. Conceptual Background
The main difference between a **Copy** and a **View** in NumPy is:
- **Copy:** Owns its own data. Changes made to the copy do not affect the original array, and vice versa.
- **View:** Does **not** own the data; it is simply another window looking at the same memory buffer. Any change to the view alters the original array!

```
import numpy as np

# Demonstrating COPY
original = np.array([1, 2, 3, 4, 5])
x = original.copy()
x[0] = 99
print('--- COPY BEHAVIOR ---')
print(f'Original array: {original}')
print(f'Copy array (modified index 0 to 99): {x}')

# Demonstrating VIEW
y = original.view()
y[0] = 777
print('\n--- VIEW BEHAVIOR ---')
print(f'Original array (affected!): {original}')
print(f'View array (modified index 0 to 777): {y}')
```

### 2. Checking Ownership with the `.base` Attribute
Every NumPy array has an attribute `base` that returns `None` if the array owns the data (copy), or references the original object if it is a view.

```
arr = np.array([10, 20, 30, 40, 50])

c = arr.copy()
v = arr.view()

print(f'Copy .base attribute: {c.base} (None means it owns its data)')
print(f'View .base attribute: {v.base} (Points to original buffer)')
print(f'Is view base identical to original? {v.base is arr}')
```
