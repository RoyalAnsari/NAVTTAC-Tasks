### 1. Conceptual Background
Sorting arranges array elements in ordered sequence. `np.sort()` returns a sorted copy of the array without modifying the original array in-place, using quicksort, mergesort, or heapsort algorithms.

```
import numpy as np

# Numeric sort
nums = np.array([3, 2, 0, 1, 9, 7, 5])
sorted_nums = np.sort(nums)
print(f'Original: {nums}')
print(f'Sorted: {sorted_nums}')

# Alphabetical sort
words = np.array(['banana', 'cherry', 'apple', 'mango'])
print(f'Sorted words: {np.sort(words)}')

# Boolean sort
bools = np.array([True, False, True, False, True])
print(f'Sorted booleans (False before True): {np.sort(bools)}')
```

### 2. Sorting 2-D Arrays along Axes and Argsort
Sorting multidimensional arrays along axis 0 (rows) or axis 1 (columns), and obtaining sort indices using `np.argsort()`.

```
matrix = np.array([[3, 2, 4], [5, 0, 1]])
print(f'Original 2D Matrix:\n{matrix}')
print(f'Sort along Axis 1 (each row sorted horizontally):\n{np.sort(matrix, axis=1)}')
print(f'Sort along Axis 0 (each column sorted vertically):\n{np.sort(matrix, axis=0)}')

# Argsort: returns indices that would sort an array
scores = np.array([88, 62, 95, 74])
sort_order = np.argsort(scores)
print(f'Scores: {scores}')
print(f'Indices of sorted order: {sort_order}')
print(f'Ordered elements: {scores[sort_order]}')
```
