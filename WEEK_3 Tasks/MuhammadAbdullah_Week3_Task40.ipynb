### 1. Conceptual Background
Random number generation is critical in Machine Learning for weight initialization, dataset splitting, stochastic gradient descent, and Monte Carlo simulations. `np.random` provides comprehensive random routines.

```
import numpy as np

np.random.seed(42)  # Seed for reproducible results

# Random integer between 0 and 100
rand_int = np.random.randint(100)
print(f'Random Integer: {rand_int}')

# 2D array of random integers
rand_matrix = np.random.randint(10, 50, size=(2, 4))
print(f'Random Matrix (2x4) between 10 and 50:\n{rand_matrix}')

# Random float between 0.0 and 1.0
rand_float = np.random.rand(5)
print(f'5 Random Floats: {rand_float}')
```

### 2. Random Choice, Shuffling, and Normal Distributions
Sampling from a finite set with probabilities, random shuffling, and Gaussian distribution generation.

```
import matplotlib.pyplot as plt

# Random choice with weighted probabilities
choices = np.random.choice(['Red', 'Green', 'Blue'], p=[0.6, 0.3, 0.1], size=10)
print(f'Weighted Random Choices: {choices}')

# Shuffling array in-place
deck = np.arange(1, 11)
np.random.shuffle(deck)
print(f'Shuffled Deck: {deck}')

# Normal (Gaussian) distribution
gaussian_samples = np.random.normal(loc=50, scale=5, size=500)
print(f'Gaussian Mean: {gaussian_samples.mean():.2f}, Std: {gaussian_samples.std():.2f}')

plt.figure(figsize=(7, 3.5))
plt.hist(gaussian_samples, bins=25, color='teal', edgecolor='black', alpha=0.7)
plt.title('Histogram of Generated Gaussian Random Variables')
plt.xlabel('Value')
plt.ylabel('Count')
plt.grid(True, alpha=0.3)
plt.show()
```
