### 1. Conceptual Background
NumPy enables vectorized conditional searching:
- `np.where(condition)`: Returns indices where the condition evaluates to True.
- `np.searchsorted(arr, val)`: Performs binary search on sorted arrays to find insertion points maintaining sorted order.

```
import numpy as np

arr = np.array([1, 2, 3, 4, 5, 4, 4, 8, 9, 10])
print(f'Array: {arr}')

# Finding indices of specific value
idx_4 = np.where(arr == 4)
print(f'Indices where value is 4: {idx_4[0]}')

# Finding indices of even numbers
idx_even = np.where(arr % 2 == 0)
print(f'Indices of even numbers: {idx_even[0]}')
print(f'Actual even values: {arr[idx_even]}')
```

### 2. Binary Search with `np.searchsorted()`
Locating insertion index in pre-sorted data efficiently in $O(\log n)$ time.

```
sorted_arr = np.array([10, 20, 30, 40, 50, 60])
print(f'Sorted Array: {sorted_arr}')

# Single value search
pos = np.searchsorted(sorted_arr, 35)
print(f'Value 35 should be inserted at index: {pos}')

# Multiple values search
multi_pos = np.searchsorted(sorted_arr, [15, 45, 65])
print(f'Values [15, 45, 65] insert indices: {multi_pos}')
```
